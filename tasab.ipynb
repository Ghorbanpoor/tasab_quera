    <br>
    همچنین اگر از کولب برای اجرای این فایل نوت‌بوک استفاده می‌کنید، قبل از ارسال فایل <code>result.zip</code>، آخرین نسخه‌ی نوت‌بوک خود را دانلود کرده و داخل فایل ارسالی قرار دهید.

```python
import os
import zipfile
import pandas

if not os.path.exists(os.path.join(os.getcwd(), 'tasab.ipynb')):
    %notebook -e tasab.ipynb

above_average_employees_df.to_csv('above_average_employees.csv', index=False)
loyal_percentage_df.to_csv('loyal_percentage.csv', index=False)
average_salary_by_quintile_df.to_csv('average_salary_by_quintile.csv', index=False)

def compress(file_names):
    print("File Paths:")
    print(file_names)
    compression = zipfile.ZIP_DEFLATED
    with zipfile.ZipFile("result.zip", mode="w") as zf:
        for file_name in file_names:
            zf.write('./' + file_name, file_name, compress_type=compression)

file_names = ['above_average_employees.csv', 'loyal_percentage.csv', 'average_salary_by_quintile.csv', 'tasab.ipynb']
compress(file_names)
```
